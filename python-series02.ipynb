# Mathematical Background
A sine wave can be described by the equation:

$$ s(t) = A \cdot \sin(2 \pi f t + \phi) $$

Where:
- $A$ = amplitude
- $f$ = frequency (Hz)
- $t$ = time (s)
- $\phi$ = phase (radians)

For this example, we use $A = 1$, $f = 1$ Hz, and $\phi = 0$, so:

$$ s(t) = \sin(2 \pi t) $$

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Create time array: 1000 points from 0 to 1 second
t = np.linspace(0, 1, 1000)

# Create sine signal: 1 Hz, amplitude 1
s = np.sin(2 * np.pi * t)

# Plot
plt.figure()
plt.plot(t, s)
plt.title('Sinusoidal function')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)

# Save plot as PNG
plt.savefig('sinusoid_draft.png')

# Show plot
plt.show()
```

## Better version: sine and cosine

Now we add a cosine wave and improve the plot with a legend and a second y-axis.

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Create time array: 1000 points from 0 to 1 second
t = np.linspace(0, 1, 1000)

# Create sine signal: 1 Hz, amplitude 1
s = np.sin(2 * np.pi * t)

# Create cosine signal: same frequency, same amplitude
c = np.cos(2 * np.pi * t)

# Create figure and main axis
fig, ax1 = plt.subplots()

# Plot sine on the main axis
ax1.plot(t, s, label='Sinus', color='tab:blue')
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Amplitude (sinus)', color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')
ax1.grid(True)

# Create a SECOND y-axis that shares the same x-axis
ax2 = ax1.twinx()

# Plot cosine on the second axis
ax2.plot(t, c, label='Cosinus', color='tab:orange')
ax2.set_ylabel('Amplitude (cosinus)', color='tab:orange')
ax2.tick_params(axis='y', labelcolor='tab:orange')

# Add title
plt.title('Sinusoidal function')

# Add legend (combine legends from both axes)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='lower left')

# Save plot as PDF (instead of PNG)
plt.savefig('sinusoid_better.pdf')

# Show the plot
plt.show()
```

## Final version: function-based approach

Now we refactor our code into a reusable function called sinusoid.

This function:
- Takes parameters for amplitude, frequency, phase, duration, and number of samples
- Returns two arrays: the time array and the signal array

This makes the code cleaner, more flexible, and easier to reuse.

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt


def sinusoid(amplitude=1.0, frequency=1.0, phase=0.0, duration=1.0, n_samples=1000):
    """
    Generate a sinusoidal signal.

    Parameters
    ----------
    amplitude : float
        Peak amplitude of the sinusoid (default: 1.0)
    frequency : float
        Frequency in Hz (default: 1.0)
    phase : float
        Phase in radians (default: 0.0)
    duration : float
        Duration in seconds (default: 1.0)
    n_samples : int
        Number of samples to generate (default: 1000)

    Returns
    -------
    t : numpy.ndarray
        Time array in seconds
    s : numpy.ndarray
        Sinusoidal signal values
    """
    # Create time array
    t = np.linspace(0, duration, n_samples)
    
    # Compute the sinusoid
    s = amplitude * np.sin(2 * np.pi * frequency * t + phase)
    
    return t, s


# Generate two sinusoids with different frequencies
t, s1 = sinusoid(amplitude=1.0, frequency=1.0)   # 1 Hz
_, s2 = sinusoid(amplitude=1.0, frequency=2.0)   # 2 Hz

# Plot both signals
fig, ax = plt.subplots()
ax.plot(t, s1, label='Sinus 1 (1 Hz)')
ax.plot(t, s2, label='Sinus 2 (2 Hz)')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.set_title('Sinusoidal function')
ax.grid(True)
ax.legend()

# Save as PDF
plt.savefig('sinusoid_final.pdf')

# Show the plot
plt.show()
```

## 3. Playing with integers

### 3.1. Even or odd

*Definition (from a mathematical point of view):*

An integer $n$ is *even* if it is divisible by 2 without a remainder:

$$n \bmod 2 = 0$$

Otherwise, $n$ is *odd*:

$$n \bmod 2 = 1$$

```python
# Create a series of numbers from 0 to 10
for n in range(11):
    if n % 2 == 0:
        print(n, "is even")
    else:
        print(n, "is odd")
```

```python
def print_even_or_odd(n):
    """
    Print whether a given integer is even or odd.

    Parameters
    ----------
    n : int
        The integer to test.

    Returns
    -------
    None
        This function prints the result but returns nothing.
    """
    if n % 2 == 0:
        print(f"{n} is even")
    else:
        print(f"{n} is odd")


# Test the function on numbers from 0 to 10
for number in range(11):
    print_even_or_odd(number)
```
